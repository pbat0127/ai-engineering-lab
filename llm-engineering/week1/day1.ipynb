```python
"""
Email subject line generator.

Week 1 exercise (LLM Engineering course), written from scratch.
Sends an email body to an OpenAI chat model with a system prompt that
constrains the output to a single subject line (max 10 words, same
language as the email). The email is wrapped in <email> tags so the
model treats it as data, not as instructions.
"""

from openai import OpenAI

client = OpenAI()

SYSTEM = ("You write email subject lines. Reply with ONE subject lien in the same language as the email, max 10 words, no quotes, no explanation.")

def suggest_subject(email: str, model: str = "gpt-6-luna") -> str: 
    resp = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content":SYSTEM},
            {"role": "user", "content": f"<email>\n{email}\n</email>"}
        ],
    )
    return resp.choices[0].message.content.strip()

email_text = """
Hola equipo,

Revisando el reporte de la versión 2.3 de la app de NovaPay, detectamos que las
transferencias programadas no se están ejecutando cuando la fecha cae en fin de semana.

El comportamiento esperado, según la historia de usuario HU-142, es que la
transferencia se procese el siguiente día hábil. Actualmente queda en estado
"pendiente" sin notificar al cliente.

Les pedimos priorizar la corrección para el próximo sprint y confirmar si el
problema también afecta a los pagos recurrentes de tarjetas.

Quedo atento a sus comentarios.

Saludos,
"""

email_text_en = """
Hi team,

Our monthly cloud bill increased by 40% in October. Most of the cost comes from
the staging environment, which runs 24/7 even though we only use it during
business hours. Could we schedule it to shut down at night and on weekends?

Thanks,
Mark
"""

print(suggest_subject(email_text))
print(suggest_subject(email_text_en))
```
